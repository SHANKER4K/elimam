```python
import pandas as pd

df = pd.read_csv('data/full_hadith.csv')
df.head()
```

```python
def fix_df(df_):
    df_ = df_[df_['language'].isin(['Arabic'])]
    df_ = df_[df_['name'].isin([
    'ara-abudawud',
    'ara-bukhari',
    'ara-dehlawi',
    'ara-ibnmajah',
    'ara-malik',
    'ara-muslim',
    'ara-nasai',
    'ara-nawawi',
    'ara-qudsi',
    'ara-tirmidhi'])]
    df_ = df_.reset_index(drop=True)
    df_ = df_.drop(columns=[
    'author',
    'direction',
    'source',
    'language',
    'has_sections',
    'Muhammad Muhyi Al-Din Abdul Hamid',
    'Shuaib Al Arnaut',
    'Zubair Ali Zai',
    'Muhammad Fouad Abd al-Baqi',
    'Salim al-Hilali',
    'Abu Ghuddah',
    'Ahmad Muhammad Shakir',
    'Bashar Awad Maarouf',])
    return df_

df = fix_df(df)
```
