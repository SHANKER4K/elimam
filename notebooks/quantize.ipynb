```python
from sentence_transformers import CrossEncoder

model = CrossEncoder(
    "NAMAA-Space/GATE-Reranker-V1",
    backend="onnx",
    trust_remote_code=True
)
model.save_pretrained(
    "models/gate_reranker_onnx",
)
```

```python
from onnxruntime.quantization import quantize_dynamic, QuantType

quantize_dynamic(
    model_input="models/gate_arabert_onnx/onnx/model.onnx",
    model_output="models/gate_arabert_onnx/onnx/model_int8.onnx",
    weight_type=QuantType.QInt8,
)
```

```python
from sentence_transformers import CrossEncoder
reranker = CrossEncoder(
    "models/gate_reranker_onnx",
    backend="onnx",
    model_kwargs={"file_name": "model_int8.onnx"}
)
```

```python
Query = 'كيف يمكن استخدام التعلم العميق في معالجة الصور الطبية؟'
Paragraph1 = 'التعلم العميق يساعد في تحليل الصور الطبية وتشخيص الأمراض'
Paragraph2 = 'الذكاء الاصطناعي يستخدم في تحسين الإنتاجية في الصناعات'

scores = reranker.rank(Query,[Paragraph1,Paragraph2])
scores
```
